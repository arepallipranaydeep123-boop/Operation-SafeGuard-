```python
#To calculate Precision, Recall, F1 Score, FCR, and FPR
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# Actual fraud labels
y_true = df['isFraud']

# Predicted fraud labels
y_pred = df['ai_fraud_flag']

# Precision
precision = precision_score(y_true, y_pred)

# Recall
recall = recall_score(y_true, y_pred)

# F1 Score
f1 = f1_score(y_true, y_pred)

# Confusion Matrix
tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()

# Fraud Capture Rate (same as Recall)
fcr = tp / (tp + fn)

# False Positive Ratio
fpr = fp / (fp + tn)

print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
print("F1 Score:", round(f1, 4))
print("FCR:", round(fcr, 4))
print("FPR:", round(fpr, 4))
```
